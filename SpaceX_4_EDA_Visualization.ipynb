# SpaceX Falcon 9 - Exploratory Data Analysis with Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("dataset_part_2.csv")
df.head()
```

## Flight Number vs. Payload Mass, colored by Class

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Payload Mass (kg)", fontsize=20)
plt.show()
```

## Flight Number vs. Launch Site

```python
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()
```

## Payload Mass vs. Launch Site

```python
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.show()
```

## Success rate of each orbit type

```python
orbit_success = df.groupby("Orbit")["Class"].mean().reset_index()
sns.barplot(x="Orbit", y="Class", data=orbit_success)
plt.xlabel("Orbit", fontsize=20)
plt.ylabel("Success Rate", fontsize=20)
plt.show()
```

## Flight Number vs. Orbit type

```python
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()
```

## Payload Mass vs. Orbit type

```python
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.show()
```

## Launch success yearly trend

```python
def Extract_year(dates):
    return [i.split("-")[0] for i in dates]

df["Year"] = Extract_year(df["Date"])
year_success = df.groupby("Year")["Class"].mean().reset_index()

sns.lineplot(x="Year", y="Class", data=year_success)
plt.xlabel("Year", fontsize=20)
plt.ylabel("Success Rate", fontsize=20)
plt.show()
```

## Feature engineering: create dummy variables for categorical columns

```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights',
    'GridFins', 'Reused', 'Legs', 'LandingPad', 'Block',
    'ReusedCount', 'Serial']]
features.head()
```

```python
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])
features_one_hot.head()
```

```python
features_one_hot = features_one_hot.astype('float64')
```

## Export to CSV

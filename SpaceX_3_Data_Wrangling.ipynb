# SpaceX Falcon 9 - Data Wrangling

Clean the data collected from Notebook 1 and create the classification label "Class" (1 = successful landing, 0 = unsuccessful landing).

```python
import pandas as pd
import numpy as np

df = pd.read_csv("dataset_part_1.csv")
df.head()
```

## Identify and calculate the percentage of missing values in each column

```python
df.isnull().sum()/len(df)*100
```

## Number of launches on each site

```python
df['LaunchSite'].value_counts()
```

## Number and occurrence of each orbit

```python
df['Orbit'].value_counts()
```

## Number and occurrence of mission outcome per orbit

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i,outcome in enumerate(landing_outcomes.keys()):
    print(i,outcome)
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1,3,5,6,7]])
bad_outcomes
```

## Create a landing outcome label from the Outcome column
We use the set of outcomes above to create a list where 0 means the booster did not land successfully, and 1 means it did.

```python
landing_class = []
for outcome in df['Outcome']:
    if outcome in bad_outcomes:
        landing_class.append(0)
    else:
        landing_class.append(1)
```

```python
df['Class']=landing_class
df[['Class']].head(8)
```

```python
df["Class"].mean()
```
